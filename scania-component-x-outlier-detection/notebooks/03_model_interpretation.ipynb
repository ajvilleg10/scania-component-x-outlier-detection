# Interpretación de outlier scores

Notebook opcional para inspeccionar predicciones por vehículo sin alterar el pipeline.

```
import csv
from pathlib import Path
RUN = Path("/content/drive/MyDrive/TFM_SCANIA/experiments/runs/debug_025")
model="lstm_autoencoder"
path=RUN/"predictions"/model/"test_vehicle_predictions.csv"
with path.open(encoding="utf-8") as f:
    rows=list(csv.DictReader(f))
rows_sorted=sorted(rows, key=lambda r: float(r["max_score"]), reverse=True)
for row in rows_sorted[:10]:
    print(row["vehicle_id"], row["max_score"], row["y_true"], row["is_outlier"])
```
