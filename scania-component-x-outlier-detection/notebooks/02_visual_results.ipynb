```
import csv
from pathlib import Path
RUN = Path("/content/drive/MyDrive/TFM_SCANIA/experiments/runs/debug_025")
comparison = RUN/"comparisons"/"comparison_all_models.csv"
with comparison.open(encoding="utf-8") as f:
    rows=list(csv.DictReader(f))
for row in rows:
    if row.get("level")=="vehicle":
        print(row.get("model"), "F1=", row.get("f1_score"), "PR-AUC=", row.get("pr_auc"))
```
