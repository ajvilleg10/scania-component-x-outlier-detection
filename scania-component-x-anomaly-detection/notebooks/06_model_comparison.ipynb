# 06 · Comparación final de modelos

Consolida las métricas finales a nivel vehículo/trayectoria y genera tablas para la sección de resultados.

```python

from pathlib import Path
import json
import numpy as np
import pandas as pd

from scania_anomaly.config import load_config, ensure_directories
from scania_anomaly.utils.reproducibility import set_global_seed

config = load_config('../config/config.yaml')
ensure_directories(config)
set_global_seed(config['project']['seed'])

DRIVE_ROOT = Path(config['paths']['drive_root'])
RAW_DIR = Path(config['paths']['raw_dir'])
PROCESSED_DIR = Path(config['paths']['processed_dir'])
OUTPUTS_DIR = Path(config['paths']['outputs_dir'])
MODELS_DIR = Path(config['paths']['models_dir'])
METRICS_DIR = Path(config['paths']['metrics_dir'])
TABLES_DIR = Path(config['paths']['tables_dir'])
print(config['project']['name'], config['project']['version'])

from pathlib import Path
from scania_anomaly.model_evaluation import compare_metric_files
```

```python

metric_files = sorted(str(p) for p in METRICS_DIR.glob('*_test_vehicle_metrics.json'))
comparison = compare_metric_files(metric_files)
comparison.to_csv(TABLES_DIR / 'model_comparison_vehicle_level.csv', index=False)
comparison
```

```python

# Orden sugerido: priorizar F1-score y PR-AUC por desbalance de clases.
ranking = comparison.sort_values(['f1_score', 'pr_auc'], ascending=False)
ranking.to_csv(TABLES_DIR / 'model_ranking_vehicle_level.csv', index=False)
ranking
```
